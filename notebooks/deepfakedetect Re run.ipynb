**Filter Images**

```python
from deepface import DeepFace
import os, shutil, random

src_real = "/kaggle/input/datasets/ayushmandatta1/deepdetect-2025/ddata/train/real"
src_fake = "/kaggle/input/datasets/ayushmandatta1/deepdetect-2025/ddata/train/fake"

def filter_asian(src_folder, out_folder, target_count):
    os.makedirs(out_folder, exist_ok=True)
    files = os.listdir(src_folder)
    random.shuffle(files)
    count = 0
    checked = 0
    for f in files:
        if count >= target_count:
            break
        checked += 1
        path = os.path.join(src_folder, f)
        try:
            result = DeepFace.analyze(path, actions=['race'], enforce_detection=False)
            dominant = result[0]['dominant_race']
            if dominant == 'asian':
                shutil.copy(path, out_folder)
                count += 1
        except:
            pass
        if checked % 20 == 0:
            print(f"Checked {checked}, found {count}")
    print(f"DONE {out_folder}: {count} images from {checked} checked")

filter_asian(src_real, "/kaggle/working/filtered_real", 500)
filter_asian(src_fake, "/kaggle/working/filtered_fake", 500)
```

**Create the Dual Model Architecture**

```python
import torch
import torch.nn as nn
from torchvision import models

class DualBranchDeepfakeDetector(nn.Module):
    def __init__(self):
        super(DualBranchDeepfakeDetector, self).__init__()
        
        # 1. Branch A (CNN) - Global Features
        self.branch_a_cnn = models.efficientnet_b0(pretrained=True)
        self.branch_a_cnn.classifier = nn.Identity() 
        
        # 2. Branch B (Eye Landmark) - Eye Biological Features
        self.branch_b_eye = models.resnet18(pretrained=True)
        self.branch_b_eye.fc = nn.Identity()
        
        # 3. Fusion Layer (Branch A + Branch B)
        combined_features_dim = 1280 + 512  
        self.fusion_layer = nn.Sequential(
            nn.Linear(combined_features_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.5)
        )
        
        # 4. Classifier (Real/Fake Label & Explanation Output)
        self.label_classifier = nn.Sequential(
            nn.Linear(256, 1),
            nn.Sigmoid() # Real / Fake Label (0 - 1)
        )
        
        self.explanation_head = nn.Linear(256, 64) # Explanation Output

    def forward(self, full_face, eye_region):
        
        feat_a = self.branch_a_cnn(full_face)
        
        
        feat_b = self.branch_b_eye(eye_region)
        
        # Fusion Layer 
        combined = torch.cat((feat_a, feat_b), dim=1)
        fused = self.fusion_layer(combined)
        
        # Classifier 
        label_output = self.label_classifier(fused)
        explanation_output = self.explanation_head(fused)
        
        return label_output, explanation_output


dual_model = DualBranchDeepfakeDetector()
print("Dual-Branch Architecture Code is Ready and Matches the Diagram!")
```

**Baseline comparison Model A**

```python
import torch
import torch.nn as nn
from torchvision import models

class SingleBranchBaseline(nn.Module):
    def __init__(self):
        super(SingleBranchBaseline, self).__init__()
        
        # Baseline, Branch A (EfficientNet)
        self.backbone = models.efficientnet_b0(pretrained=True)
        
        # Binary classification (Real vs Fake)
        num_ftrs = self.backbone.classifier[1].in_features
        self.backbone.classifier = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(num_ftrs, 1),
            nn.Sigmoid()
        )

    def forward(self, full_face):
        
        return self.backbone(full_face)

baseline_model = SingleBranchBaseline()
print("Baseline Model (Branch A in isolation) Created Successfully!")
```

**Stratified 5-Fold CV, F1-score, AUC-ROC, paired t-test**

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, roc_auc_score
from scipy.stats import ttest_rel

# Stratified 5-Fold Cross-Validation Setup
def get_stratified_cv(n_splits=5):
    
    return StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Evaluation Metrics Calculation
ftl_score_metric = lambda y_true, y_pred_labels: f1_score(y_true, y_pred_labels)
auc_roc_metric = lambda y_true, y_pred_probs: roc_auc_score(y_true, y_pred_probs)

def evaluate_predictions(y_true, y_pred_probs, y_pred_labels):
   
    f1 = f1_score(y_true, y_pred_labels, zero_division=0)
    try:
        auc = roc_auc_score(y_true, y_pred_probs)
    except ValueError:
        auc = 0.5 
    return {"F1-Score": f1, "AUC-ROC": auc}

#  Paired t-test for Statistical Significance
def perform_paired_ttest(baseline_scores, dual_branch_scores):
   
    t_stat, p_value = ttest_rel(dual_branch_scores, baseline_scores)
    return {"t-statistic": t_stat, "p-value": p_value}

print("Evaluation pipeline functions defined successfully!")
```

**PREPROCESSING PIPELINE (Face Detection + Compression + Blur)**

```python
!pip install mtcnn

import cv2
import numpy as np
import os
from mtcnn import MTCNN

detector = MTCNN()

#Face Detection and Cropping
def detect_and_crop_face(image):
    result = detector.detect_faces(image)
    if result:
        x, y, w, h = result[0]['box']
        x, y = max(0, x), max(0, y)
        face = image[y:y+h, x:x+w]
        return cv2.resize(face, (224, 224))
    return None

# Compression and Blur functions
def apply_jpeg_compression(image, quality=60):
    encode_param = [int(cv2.IMWRITE_JPEG_QUALITY), quality]
    result, encimg = cv2.imencode('.jpg', image, encode_param)
    decimg = cv2.imdecode(encimg, 1)
    return decimg

def apply_gaussian_blur(image, kernel_size=(3, 3)):
    return cv2.GaussianBlur(image, kernel_size, 0)

#Full pipeline
def process_single_image(image_path):
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    face = detect_and_crop_face(img_rgb)
    if face is None:
        return None 

    img = cv2.cvtColor(face, cv2.COLOR_RGB2BGR)

    if np.random.rand() > 0.5:
        q = np.random.randint(40, 90)
        img = apply_jpeg_compression(img, quality=q)

    if np.random.rand() > 0.7:
        img = apply_gaussian_blur(img, kernel_size=(3, 3))

    img = img.astype('float32') / 255.0

    return img

print("Full preprocessing pipeline ready (face detection + compression + blur)")
```

**Eye region extract**

```python
import cv2
import numpy as np
from mtcnn import MTCNN

detector = MTCNN()

def extract_eye_region(image_path):
    
    img = cv2.imread(image_path)
    if img is None:
        print(f"Error: Could not read image from {image_path}")
        return None
        
    # Convert BGR to RGB since MTCNN expects RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    results = detector.detect_faces(img_rgb)
    
    if results:
        keypoints = results[0]['keypoints']
        left_eye = keypoints['left_eye']
        right_eye = keypoints['right_eye']
        
        # Calculate bounding box covering both eyes with a safety margin
        min_x = min(left_eye[0], right_eye[0]) - 30
        max_x = max(left_eye[0], right_eye[0]) + 30
        min_y = min(left_eye[1], right_eye[1]) - 25
        max_y = max(left_eye[1], right_eye[1]) + 25
        
        h_img, w_img, _ = img_rgb.shape
        min_x, min_y = max(0, min_x), max(0, min_y)
        max_x, max_y = min(w_img, max_x), min(h_img, max_y)
        
        eye_crop = img_rgb[min_y:max_y, min_x:max_x]
        if eye_crop.size == 0:
            
            return cv2.resize(img_rgb, (224, 224))
            
       
        return cv2.resize(eye_crop, (224, 224))
        
    else:
        print(f"Warning: No face/eyes detected in {image_path}")
        return None
```

```python
import numpy as np
import os

filtered_real = "/kaggle/working/filtered_real"
filtered_fake = "/kaggle/working/filtered_fake"

faces, eyes, labels = [], [], []

def build_dataset(folder, label):
    files = os.listdir(folder)
    kept, skipped = 0, 0
    for fname in files:
        path = os.path.join(folder, fname)

        face_img = process_single_image(path)          # face branch input
        eye_img  = extract_eye_region(path)             # eye branch input

        if face_img is None or eye_img is None:
            skipped += 1
            continue

        faces.append(face_img)
        eyes.append(eye_img.astype("float32") / 255.0)
        labels.append(label)
        kept += 1

    print(f"{folder}: kept {kept}, skipped {skipped}")

build_dataset(filtered_real, 1)   # 1 = Real
build_dataset(filtered_fake, 0)   # 0 = Fake

faces = np.array(faces, dtype="float32")
eyes  = np.array(eyes, dtype="float32")
labels = np.array(labels, dtype="float32")

print("Final shapes:", faces.shape, eyes.shape, labels.shape)

np.save("/kaggle/working/faces.npy", faces)
np.save("/kaggle/working/eyes.npy", eyes)
np.save("/kaggle/working/labels.npy", labels)
```

**Pytorch dataset class**

```python
import torch
from torch.utils.data import Dataset, DataLoader

class DeepfakeDataset(Dataset):
    def __init__(self, faces, eyes, labels, indices):
        self.faces = faces[indices]
        self.eyes = eyes[indices]
        self.labels = labels[indices]

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        face = torch.tensor(self.faces[idx]).permute(2, 0, 1)
        eye  = torch.tensor(self.eyes[idx]).permute(2, 0, 1)
        label = torch.tensor(self.labels[idx], dtype=torch.float32)
        return face, eye, label

print("Dataset class ready.")
```

**Training and Evaluation function**

```python
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

def train_one_model(model, train_loader, is_dual, epochs=10, lr=1e-4):
    model = model.to(device)
    model.train()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.BCELoss()

    for epoch in range(epochs):
        total_loss = 0.0
        for face, eye, label in train_loader:
            face, eye, label = face.to(device), eye.to(device), label.to(device).unsqueeze(1)
            optimizer.zero_grad()

            if is_dual:
                output, _ = model(face, eye)
            else:
                output = model(face)

            loss = criterion(output, label)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        print(f"  Epoch {epoch+1}/{epochs} - Loss: {total_loss/len(train_loader):.4f}")

    return model


def evaluate_one_model(model, val_loader, is_dual):
    model.eval()
    y_true, y_pred_probs = [], []

    with torch.no_grad():
        for face, eye, label in val_loader:
            face, eye = face.to(device), eye.to(device)

            if is_dual:
                output, _ = model(face, eye)
            else:
                output = model(face)

            y_pred_probs.extend(output.cpu().numpy().flatten().tolist())
            y_true.extend(label.numpy().flatten().tolist())

    y_pred_probs = np.array(y_pred_probs)
    y_true = np.array(y_true)
    y_pred_labels = (y_pred_probs >= 0.5).astype(int)

    return y_true, y_pred_probs, y_pred_labels

print("Training/evaluation functions ready.")
```

**Main 5-Fold CV Training Loop**

```python
from sklearn.model_selection import StratifiedKFold

cv = get_stratified_cv(n_splits=5)

baseline_f1_scores, baseline_auc_scores = [], []
dual_f1_scores, dual_auc_scores = [], []

fold_num = 1
for train_idx, val_idx in cv.split(faces, labels):
    print(f"\n========== FOLD {fold_num}/5 ==========")

    train_ds = DeepfakeDataset(faces, eyes, labels, train_idx)
    val_ds   = DeepfakeDataset(faces, eyes, labels, val_idx)
    train_loader = DataLoader(train_ds, batch_size=16, shuffle=True)
    val_loader   = DataLoader(val_ds, batch_size=16, shuffle=False)

    print("Training baseline...")
    baseline = SingleBranchBaseline()
    baseline = train_one_model(baseline, train_loader, is_dual=False, epochs=10)
    y_true, y_probs, y_labels = evaluate_one_model(baseline, val_loader, is_dual=False)
    b_metrics = evaluate_predictions(y_true, y_probs, y_labels)
    baseline_f1_scores.append(b_metrics["F1-Score"])
    baseline_auc_scores.append(b_metrics["AUC-ROC"])
    print("Baseline fold metrics:", b_metrics)

    print("Training dual-branch...")
    dual = DualBranchDeepfakeDetector()
    dual = train_one_model(dual, train_loader, is_dual=True, epochs=10)
    y_true, y_probs, y_labels = evaluate_one_model(dual, val_loader, is_dual=True)
    d_metrics = evaluate_predictions(y_true, y_probs, y_labels)
    dual_f1_scores.append(d_metrics["F1-Score"])
    dual_auc_scores.append(d_metrics["AUC-ROC"])
    print("Dual-branch fold metrics:", d_metrics)

    fold_num += 1

print("\n\nAll folds complete.")
```

```python
import numpy as np, pandas as pd, json, torch, matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score

SEED = 42
torch.manual_seed(SEED); np.random.seed(SEED)
# Label convention used everywhere: 1 = Real, 0 = Fake

cv = get_stratified_cv(n_splits=5)
rows = []                      # one row per image, out-of-fold
fold_scores = {"base_f1":[], "base_auc":[], "dual_f1":[], "dual_auc":[]}

for fold, (tr, va) in enumerate(cv.split(faces, labels), start=1):
    print(f"===== FOLD {fold}/5 =====")
    tr_loader = DataLoader(DeepfakeDataset(faces, eyes, labels, tr), batch_size=16, shuffle=True)
    va_loader = DataLoader(DeepfakeDataset(faces, eyes, labels, va), batch_size=16, shuffle=False)

    base = train_one_model(SingleBranchBaseline(), tr_loader, is_dual=False, epochs=10)
    yt, pb, lb = evaluate_one_model(base, va_loader, is_dual=False)
    mb = evaluate_predictions(yt, pb, lb)

    dual = train_one_model(DualBranchDeepfakeDetector(), tr_loader, is_dual=True, epochs=10)
    yt2, pd_, ld = evaluate_one_model(dual, va_loader, is_dual=True)
    md_ = evaluate_predictions(yt2, pd_, ld)

    fold_scores["base_f1"].append(mb["F1-Score"]); fold_scores["base_auc"].append(mb["AUC-ROC"])
    fold_scores["dual_f1"].append(md_["F1-Score"]); fold_scores["dual_auc"].append(md_["AUC-ROC"])
    for i, idx in enumerate(va):
        rows.append(dict(index=int(idx), fold=fold, label=int(yt[i]),
                         base_prob=float(pb[i]), dual_prob=float(pd_[i])))

oof = pd.DataFrame(rows).sort_values("index").reset_index(drop=True)
oof["base_pred"] = (oof.base_prob >= 0.5).astype(int)
oof["dual_pred"] = (oof.dual_prob >= 0.5).astype(int)
oof.to_csv("/kaggle/working/oof_predictions.csv", index=False)
json.dump(fold_scores, open("/kaggle/working/final_results_rerun.json","w"), indent=2)
print({k: round(float(np.mean(v)),4) for k,v in fold_scores.items()})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 3.6))
summary = {}
for ax, (name, col) in zip(axes, [("Baseline","base_pred"), ("Dual-branch","dual_pred")]):
    cm = confusion_matrix(oof.label, oof[col], labels=[0,1])      # rows = true (Fake, Real)
    tn, fp, fn, tp = cm.ravel()   # here "positive" = Real
    summary[name] = dict(fake_as_real=int(fp), real_as_fake=int(fn),
                         fake_recall=round(tn/(tn+fp),4), real_recall=round(tp/(tp+fn),4))
    ax.imshow(cm, cmap="Blues")
    ax.set_xticks([0,1]); ax.set_xticklabels(["Pred Fake","Pred Real"])
    ax.set_yticks([0,1]); ax.set_yticklabels(["True Fake","True Real"])
    for (i,j), v in np.ndenumerate(cm): ax.text(j, i, v, ha="center", va="center", fontsize=12)
    ax.set_title(name)
plt.tight_layout(); plt.savefig("/kaggle/working/confusion_matrices.png", dpi=200); plt.show()
print(json.dumps(summary, indent=2))


plt.figure(figsize=(4.2,4))
for name, col in [("Baseline","base_prob"), ("Dual-branch","dual_prob")]:
    fpr, tpr, _ = roc_curve(oof.label, oof[col])
    plt.plot(fpr, tpr, label=f"{name} (AUC={roc_auc_score(oof.label, oof[col]):.3f})")
plt.plot([0,1],[0,1],"k--",lw=0.8); plt.xlabel("False positive rate"); plt.ylabel("True positive rate")
plt.legend(loc="lower right"); plt.tight_layout(); plt.savefig("/kaggle/working/roc_curves.png", dpi=200); plt.show()
```

```python
oof["base_wrong"] = oof.base_pred != oof.label
oof["dual_wrong"] = oof.dual_pred != oof.label
both = oof[oof.base_wrong & oof.dual_wrong]
only_base = oof[oof.base_wrong & ~oof.dual_wrong]
only_dual = oof[~oof.base_wrong & oof.dual_wrong]
print("wrong in BOTH models:", len(both), "| only baseline wrong:", len(only_base), "| only dual wrong:", len(only_dual))
print("errors per fold (baseline / dual):")
print(oof.groupby("fold")[["base_wrong","dual_wrong"]].sum())

def grid(df, title, fname, n=10):
    df = df.sort_values("dual_prob").head(n) if len(df) else df
    if len(df)==0: print("none for", title); return
    fig, axs = plt.subplots(2, 5, figsize=(10,4.4))
    for ax in axs.ravel(): ax.axis("off")
    for ax, (_, r) in zip(axs.ravel(), df.iterrows()):
        ax.imshow(faces[int(r["index"])])
        ax.set_title(f"true={'Real' if r.label==1 else 'Fake'}\nbase={r.base_prob:.2f} dual={r.dual_prob:.2f}", fontsize=7)
    plt.suptitle(title); plt.tight_layout(); plt.savefig(fname, dpi=160); plt.show()

grid(both, "Misclassified by BOTH models", "/kaggle/working/misclassified_both.png")
grid(only_dual, "Misclassified ONLY by dual-branch", "/kaggle/working/misclassified_only_dual.png")
```

```python
d_f1  = np.array(fold_scores["dual_f1"])  - np.array(fold_scores["base_f1"])
d_auc = np.array(fold_scores["dual_auc"]) - np.array(fold_scores["base_auc"])
print("full mean diff   F1: %.4f  AUC: %.4f" % (d_f1.mean(), d_auc.mean()))
for k in range(5):
    keep = [i for i in range(5) if i != k]
    print(f"without fold {k+1}: F1 {d_f1[keep].mean():+.4f}  AUC {d_auc[keep].mean():+.4f}")


from sklearn.metrics import f1_score
best = {}
for name, col in [("Baseline","base_prob"), ("Dual-branch","dual_prob")]:
    ths = np.linspace(0.05, 0.95, 91)
    f1s = [f1_score(oof.label, (oof[col]>=t).astype(int)) for t in ths]
    best[name] = (round(float(ths[int(np.argmax(f1s))]),2), round(float(max(f1s)),4),
                  round(float(f1_score(oof.label,(oof[col]>=0.5).astype(int))),4))
print("(best threshold, best F1, F1 at 0.5):", best)   # exploratory only, tuned on the same data
```
